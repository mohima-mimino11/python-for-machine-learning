### FILE READ

```python
file = open("./sample.txt", "r")
content = file.read()
print(content)
print(type(content))

file.close()
print(file.closed)  
```

```python
file = open("./sample.txt", "r")

content = file.readlines()

content = list(map(str.strip, content))

# print(content)

filterContent = list(filter(lambda x : len(x) >= 0, content))

print(filterContent)
```

```python
# correct and reasonable syntax for file handling as it closes file automatically and go on with the rest of the program
with open("./sample.txt", "r") as file:
  content = file.readlines()
  print(content)
  
print(file.closed)
```

```python
with open("./sample.txt","r") as file:
  for line in file:
    l = line.strip()
  print(l)
```

### FILE WRITE

```python
# creating a new file and writing there
with open("./test.txt", "w") as file:
  file.write("Hello test file\n")
  file.write("How is learning going on?")
```

```python

# writing on a existing file
# it overwrites the file

with open("./test.txt", "w") as file:
  file.write("Second Attempt\n")
  file.write("idk what I am doin buddy!\n")
  
```

```python
# Appending on a existing file without overwriting
with open("./test.txt", "a") as file:
  file.write("Blah Blah\n")
  file.write("Listen, Morty, I hate to break it to you, but what people call “love” is just a chemical reaction that compels animals to breed.\n")
  file.write("It hits hard, Morty, then it slowly fades, leaving you stranded in a failing marriage. I did it. Your parents are gonna do it. Break the cycle, Morty. Rise above. Focus on science.\n")
  
```

```python
#  Adding a list in a file
strings = ["Wubba", "Lubba", "Dub", "Dub"]
with open("./test2.txt", "w") as file:
  file.writelines(strings)
```

### File Pointer/Cursor

```python
# cursor postion in the file
with open("./sample.txt", "r") as file:
  print(file.tell())
  print(file.read())
  print(file.tell())
```

```python
# cursor postion in the file
with open("./sample.txt", "r") as file:
  print(file.tell())
  print(file.read(5))
  print(file.tell())
  print(file.read())
  print(file.tell())
  
```

```python
# cursor postion in the file seek method for getting the pointer forward and reverse both
with open("./sample.txt", "r") as file:
  print(file.tell())
  print(file.read(5))
  print(file.tell())
  file.seek(0)
  print(file.tell())
  print(file.read(5))
  print(file.tell())
  
  
```
