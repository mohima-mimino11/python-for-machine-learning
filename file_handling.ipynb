```python
file = open("./sample.txt", "r")

content = file.readlines()

content = list(map(str.strip, content))

# print(content)

filterContent = list(filter(lambda x : len(x) >= 0, content))

print(filterContent)
```

```python
# correct and reasonable syntax for file handling as it closes file automatically and go on with the rest of the program
with open("./sample.txt", "r") as file:
  content = file.readlines()
  print(content)
  
print(file.closed)
```

### File Pointer/Cursor

```python
# cursor postion in the file
with open("./sample.txt", "r") as file:
  print(file.tell())
  print(file.read())
  print(file.tell())
```

```python
# cursor postion in the file
with open("./sample.txt", "r") as file:
  print(file.tell())
  print(file.read(5))
  print(file.tell())
  print(file.read())
  print(file.tell())
  
```

```python
# cursor postion in the file seek method for getting the pointer forward and reverse both
with open("./sample.txt", "r") as file:
  print(file.tell())
  print(file.read(5))
  print(file.tell())
  file.seek(0)
  print(file.tell())
  print(file.read(5))
  print(file.tell())
  
  
```

### Practice Problems

```python
cleaned_list = []
with open("emails.txt", "r") as file:
  lines = file.readlines()
  for l in lines:
    space_clean_line = l.strip()
    lowercase_line = space_clean_line.lower()
    cleaned_line = lowercase_line.replace(" ", "")
    if cleaned_line:
      cleaned_list.append(cleaned_line)
  
unique_emails = list(set(cleaned_list))
unique_emails.sort()
with open("unique_emails.txt", "w") as output:
  for email in unique_emails:
    output.write(email + "\n")

print(unique_emails)
    
```

```python
cleaned_names = []
with open("names.txt", "r") as file:
  lines = file.readlines()
  for name in lines:
    clean_name = name.strip().capitalize()
    if clean_name:
      cleaned_names.append(clean_name)
  
unique_names = list(set(cleaned_names))
unique_names.sort()
with open("unique_names.txt", "w") as output:
  for name in unique_names:
    output.write(name + "\n")
print(unique_names)
```

### Practice Problems

```python
a_friends = {"Rahim", "Karim", "Sakib", "Jamal"}
b_friends = {"Sakib", "Jamal", "Rafiq", "Nadim"}
mutual_friends = a_friends.intersection(b_friends)
print(mutual_friends)
a_unique_friends = a_friends.difference(b_friends)
print(f"Unique to A: {a_unique_friends}")
b_unique_friends = b_friends.difference(a_friends)
print(f"Unique to B: {b_unique_friends}")

total_unique_friends = len(mutual_friends) + len(a_unique_friends) + len(b_unique_friends)
print(f"Total unique friends: {total_unique_friends}")
```

```python
students = [("Rafi", 89), ("Sumi", 95), ("Hasan", 90), ("Nila", 75), ("Anik", 98)]
top_students = []
for student in students:
  if student[1] >= 90:
    top_students.append(student)

sorted_toppers = sorted(top_students, key=lambda x: x[1], reverse=True)
print(f"""
  Top 3 students:\n
  {sorted_toppers[0][0]}-{sorted_toppers[0][1]}\n
  {sorted_toppers[1][0]}-{sorted_toppers[1][1]} \n
  {sorted_toppers[2][0]}-{sorted_toppers[2][1]}
  """)
```

```python
words = ["sun", "planet", "moon", "star", "universe"]
filtered_words = list(filter(lambda x: len(x) > 4, words))
print(filtered_words)
```
