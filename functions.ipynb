### Parameters

```python
def greet(user="guest"):
  print(f"hello {user}")

greet("mimino")
greet()
```

```python
def squareAddition(a, b, c):
  print(f"{a} {b} {c}")
  a = a ** 2
  b = b ** 2
  return a + b + c

# positional arguments
ans = squareAddition(10, 20, 30) + 10
print(ans)
# Keyword arguments
ans = squareAddition(b = 10 , c = 20, a = 30) + 10
print(ans)
```

```python
def sqaureAddition(*args): # for taking variable parameters
  summation = 0
  for i in args:
    i = i ** 2
    summation += i
  return summation

ans = sqaureAddition(10,20, 30, 40, 50)
print(ans)
```

```python
def student(**kwargs): # for taking dictionary variable parameters
  for key,val in kwargs.items():
    print(f"{key} : {val}")
    
student(name = "mimino" , cls =10 , roll=30 , marks = 30.5) 
```

```python
# Default parameters
def power(base, exponent=2):
  return base ** exponent

print(power(3))      # Output: 9 (3^2)
print(power(3, 3))   # Output: 27 (3^3)

# Keyword arguments
def introduce(name, age, city):
  return f"{name} is {age} years old and lives in {city}"

print(introduce(age=25, city="New York", name="Alice"))
# Output: Alice is 25 years old and lives in New York
```

### generator

```python
lst = [x for x in range(500)]
# print(lst)
def dataLoader(chunkSize,lst):
  for i in range(0, len(lst), chunkSize):
    yield lst[i:i+chunkSize]
    
x = dataLoader(5, lst)
print(type(x))
print(next(x))
print(next(x))
print(next(x))
```

### LAMBDA Functions


```python
def squareAddition(x,y):
  return x ** 2 + y ** 2

print(squareAddition(2,3))

sqaureAdd = lambda x,y : x ** 2 + y ** 2

print(sqaureAdd(2, 3))
  
```

```python
x = lambda a, b : a - b
print(x(5, 3))
```

```python
def myfunc(n):
  return lambda a : a * n

mydoubler = myfunc(2)
mytripler = myfunc(3)

print(mydoubler(11))
print(mytripler(11))
```

```python
# Lambda functions are commonly used with built-in functions like map(), filter(), and sorted().
# The map() function applies a function to every item in an iterable:
numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
```

```python
# The filter() function creates a list of items for which a function returns True:
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
odd_numbers = list(filter(lambda x: x % 2 != 0, numbers))
print(odd_numbers)
```

```python
# The sorted() function can use a lambda as a key for custom sorting:
students = [("Emil", 25), ("Tobias", 22), ("Linus", 28)]
sorted_students = sorted(students, key=lambda x: x[1])
print(sorted_students)
```

```python
words = ["apple", "pie", "banana", "cherry"]
sorted_words = sorted(words, key=lambda x: len(x))
print(sorted_words)
```

```python
# Anonymous functions
square = lambda x: x ** 2
print(square(5))  # Output: 25

# Common use: with map, filter
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x**2, numbers))
print(squared)  # Output: [1, 4, 9, 16, 25]

evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)  # Output: [2, 4]

# reduce: combine iterable into one value (from functools)
from functools import reduce

product = reduce(lambda acc, x: acc * x, numbers, 1)
print(product)  # Output: 120 (1*2*3*4*5)

product2 = reduce(lambda acc, x: acc * x, numbers)
print(product2)  # Output: 120 (1*2*3*4*5)
```

### G. Max and MIN

```python
from functools import reduce
n = int(input())

lst = list(map(int, input().split()))
# print(lst)

maxValue = reduce(lambda x , y : x if x > y else y, lst)
minValue = reduce(lambda x , y : x if x < y else y, lst)
print(minValue, maxValue)
```

### J. Average 

```python
from functools import reduce
n = int(input())

lst = list(map(float, input().split()))
# print(lst)
avg = reduce(lambda x, y : x+y , lst) / len(lst)

print(avg)
```
