# String Methods

```python
string = "I wish you well in Hell! haha! yes hell! haha!"
print(len(string))
processedString = string.lower()
#To find substring
print("well" in processedString)
print("Well" in processedString) #As python is case-sensitive

# To find index of substring
forwardIndex = processedString.find("hell")
print(forwardIndex)
backwardIndex = processedString.rfind("hell")
print(backwardIndex)

# For counting substring ocurrences
count = processedString.count("hell")
print(count)

string2 = string.replace("well", "jail")
print(string)
print(string2)
```

# List and Lists Methods

```python
# declaration
numbers = [10, 40, 70, 30, 20]
print(numbers)
# access
print(numbers[3])
numbers[3] = 50
print(numbers[3])
newList = numbers[0:3]
print(numbers)
print(newList)

## modifications 

## adding a item 
numbers.append(80)
print(numbers)


## insertion at any position 
numbers.insert(0,45)
print(numbers)


## deleting the last value 
numbers.pop()

print(numbers)

## deleting a specific values 
numbers.remove(40)

print(numbers)

# ascending sort
numbers.sort()
print(numbers)
# reversing the ascending sort
numbers.reverse()
print(numbers)
```

```python
# descending sort
numbers.sort(reverse=True)

print(numbers)
```

```python
text = "yoo brother"

# Basic slicing
print(text[0:3])    # Output: yoo (indices 0, 1, 2)
print(text[4:11])   # Output: brother (indices 4 to 10)
print(text[:3])     # Output: yoo (from start to index 2)
print(text[4:])     # Output: brother (from index 4 to end)
print(text[:])      # Output: yoo brother (entire string)

# With step
print(text[0:11:2])  # Output: yo rte (every 2nd character)
print(text[::2])     # Output: yo rte (every 2nd character from start to end)
print(text[::-1])    # Output: rehtorb ooy (reverse string)
```

```python
# Creating lists
fruits = ["apple", "banana", "cherry"]
numbers = [1, 2, 3, 4, 5]

# Accessing elements
print(fruits[0])      # Output: apple (first element)
print(fruits[-1])     # Output: cherry (last element)

# Slicing
print(numbers[1:3])   # Output: [2, 3] (index 1 to 2)
print(numbers[:3])    # Output: [1, 2, 3] (first 3)
print(numbers[2:])    # Output: [3, 4, 5] (from index 2)

# Modifying lists
fruits.append("orange")      # Add to end
fruits.insert(1, "grape")    # Insert at index
fruits.remove("banana")     # Remove element
fruits.pop()                 # Remove last element

# List comprehension
squares = [x**2 for x in range(5)]
print(squares)  # Output: [0, 1, 4, 9, 16]

# With condition
evens = [x for x in range(10) if x % 2 == 0]
print(evens)  # Output: [0, 2, 4, 6, 8]
```

### Dictionaries

```python
student = {
  "name" : "Alice",
  "age" : 20,
  "grades" : [85, 90, 88]
}

# accessing values
print(student["name"])
print(student.get("age"))
print(student.get("city", "N/A")) # Output: N/A (default if key doesn't exist

# modifying dictionaries
student["city"] = "New York"
student["age"] = 21
del student["grades"]

print(student["name"])
print(student.get("age"))
print(student.get("city", "N/A"))

# Iterating
for key, value in student.items():
  print(f"{key}:{value}")

# Dictionary Comprehension
squares_dict = {x: x**2 for x in range(5)}
print(squares_dict)
```
